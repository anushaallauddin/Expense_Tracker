```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("train.csv")

df.head()
```

```python
# Check the structure and data types
df.info()
```

```python
# Get statistical summary
df.describe()
```

```python
# Check missing values in each column
df.isnull().sum()
```

```python
# Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with the most common value
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Drop Cabin because too many values are missing
df = df.drop(columns=["Cabin"])

# Check missing values again
df.isnull().sum()
```

```python
# Chart 1: Survival Count

df["Survived"].value_counts().plot(kind="bar")

plt.title("Number of Passengers by Survival Status")
plt.xlabel("Survival Status (0 = Not Survived, 1 = Survived)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
# Chart 2: Age Distribution

df["Age"].plot(kind="hist", bins=20)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

The Titanic dataset contains information about passengers, including their age, gender, passenger class, fare, and survival status. The survival chart shows that more passengers did not survive than survived. The age distribution shows that most passengers were adults, with fewer very young and elderly passengers. Overall, the analysis provides a clear overview of passenger characteristics and survival outcomes.


```python
# Save the cleaned dataset

df.to_csv("titanic_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
